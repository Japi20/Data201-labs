## 9 — Read and clean Auto

```python
Auto = pd.read_csv(data_folder + "Auto.csv")
print("Original dimensions:", Auto.shape)
Auto.head()
```

As in the lab, I replace the literal `"?"` in horsepower with a missing value and convert the column to numeric. `dropna()` then removes rows with missing data, as the exercise requires. The counts before and after make the removal visible.

```python
Auto["horsepower"] = Auto["horsepower"].replace("?", pd.NA)
Auto["horsepower"] = pd.to_numeric(Auto["horsepower"])
Auto.isna().sum()
```

```python
Auto = Auto.dropna()
print("Cleaned dimensions:", Auto.shape)
Auto.isna().sum()
```

Five rows with missing horsepower are removed, leaving **392 rows and 9 variables**. Subsequent references to the full Auto dataset mean these 392 complete observations.

## 9(a) — Quantitative and qualitative variables

**Quantitative:** mpg, cylinders, displacement, horsepower, weight, acceleration, and year. Cylinders is a discrete count; year is an ordered time variable, recorded with two digits. Both can also be grouped for plots without changing what they measure.

**Qualitative:** origin and name. Origin's codes 1, 2, and 3 mean USA, Europe, and Japan, as in the lab. Name identifies the vehicle/model.

```python
quantitative = ["mpg", "cylinders", "displacement", "horsepower", "weight", "acceleration", "year"]
```

## 9(b) — Ranges

```python
auto_ranges = pd.DataFrame({
    "Minimum": Auto[quantitative].min(),
    "Maximum": Auto[quantitative].max(),
    "Range width": Auto[quantitative].max() - Auto[quantitative].min()
})
auto_ranges
```

## 9(c) — Means and standard deviations

```python
auto_summary = pd.DataFrame({
    "Mean": Auto[quantitative].mean(),
    "Sample SD": Auto[quantitative].std()
})
auto_summary
```

## 9(d) — Remove observations 10 through 85

I interpret these as positions in the **cleaned dataset**, in its current order. Python starts at position 0 and excludes a slice's upper endpoint, so the 10th–85th observations correspond to 9:85: 76 rows.

Auto.index[9:85] obtains those rows' labels; drop() removes those labels. Saving the result as Auto_subset preserves the full cleaned Auto for parts (e) and (f). The row labels are not assumed to match positions after missing rows were removed.

```python
removed_labels = Auto.index[9:85]
Auto_subset = Auto.drop(removed_labels)
print("Removed rows:", len(removed_labels))
print("Remaining dimensions:", Auto_subset.shape)
```

```python
subset_summary = pd.DataFrame({
    "Minimum": Auto_subset[quantitative].min(),
    "Maximum": Auto_subset[quantitative].max(),
    "Range width": Auto_subset[quantitative].max() - Auto_subset[quantitative].min(),
    "Mean": Auto_subset[quantitative].mean(),
    "Sample SD": Auto_subset[quantitative].std()
})
subset_summary
```

There are **316 observations** left. These numerical summaries apply to the quantitative variables. For the qualitative predictors, the following frequency tables describe their categories; means and standard deviations of vehicle names or origin codes would not be meaningful.
